```python
import xarray as xr
import xesmf as xe
```

```python
siteid = "01205500"
```

# Step 1: read in the targeted domain file for the model

```python
ds = xr.open_dataset("namerica_domain.%s.nc"%siteid)
```

# Step 2: read in the meteorological forcing at original spatial resolution

```python
ds_met = xr.open_dataset("ERA5.01205500.orig_rsln.2012.nc")
ds_met = ds_met.rename({"latitude":"lat","longitude":"lon"})
```

```python
ds_met.VP.mean(dim='time').plot()
```

# Step 3: Create a regridder (A mapping between the original grid and the targeted grid)

```python
regridder = xe.Regridder(ds_met,ds["mask"],method='bilinear')
```

# Step 4: Conduct the regridding

```python
ds_met_regrid = regridder(ds_met)
```

```python
ds_met_regrid.VP.mean(dim='time').plot()
```

```python
ds_met_regrid.to_netcdf("ERA5.01205500.regrid.2012.nc")
```
